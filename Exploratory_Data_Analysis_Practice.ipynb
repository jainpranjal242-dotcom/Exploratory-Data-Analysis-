This educational project uses illustrative synthetic student-performance data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
raw = pd.read_csv('student_performance_raw.csv')
display(raw.head())
```

## 1. Dataset structure

```python
print('Shape:', raw.shape)
print('Columns:', raw.columns.tolist())
print(raw.dtypes)
raw.info()
display(raw.describe())
```

## 2. Missing values and duplicates

```python
print(raw.isna().sum())
print('Duplicate rows:', raw.duplicated().sum())
```

## 3. Clean the data

Remove exact duplicates and fill the missing study-hours value with the median.

```python
df = raw.drop_duplicates().copy()
df['Study_Hours'] = df['Study_Hours'].fillna(df['Study_Hours'].median())
print('Rows before:', len(raw), '| rows after:', len(df))
print('Missing values after cleaning:\n', df.isna().sum())
df.to_csv('student_performance_cleaned.csv', index=False)
```

## 4. Summary statistics

```python
display(df.describe(include='all'))
print('Average exam score:', round(df['Exam_Score'].mean(), 2))
print('Median exam score:', df['Exam_Score'].median())
```

## 5. Visualization 1 — score distribution

```python
plt.figure(figsize=(7,4))
plt.hist(df['Exam_Score'], bins=7, edgecolor='black')
plt.title('Distribution of Exam Scores')
plt.xlabel('Exam Score'); plt.ylabel('Number of Students')
plt.tight_layout(); plt.show()
```

## 6. Visualization 2 — study hours vs score

```python
plt.figure(figsize=(7,4))
plt.scatter(df['Study_Hours'], df['Exam_Score'])
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours'); plt.ylabel('Exam Score')
plt.grid(True, alpha=0.25); plt.tight_layout(); plt.show()
print('Correlation:', round(df['Study_Hours'].corr(df['Exam_Score']), 3))
```

## 7. Visualization 3 — average score by study method

```python
means = df.groupby('Study_Method')['Exam_Score'].mean().sort_values(ascending=False)
plt.figure(figsize=(7,4))
plt.bar(means.index, means.values)
plt.title('Average Exam Score by Study Method')
plt.xlabel('Study Method'); plt.ylabel('Average Exam Score')
plt.xticks(rotation=15); plt.tight_layout(); plt.show()
display(means)
```

## 8. Visualization 4 — attendance vs score

```python
plt.figure(figsize=(7,4))
plt.scatter(df['Attendance_Percent'], df['Exam_Score'])
plt.title('Attendance vs Exam Score')
plt.xlabel('Attendance (%)'); plt.ylabel('Exam Score')
plt.grid(True, alpha=0.25); plt.tight_layout(); plt.show()
print('Correlation:', round(df['Attendance_Percent'].corr(df['Exam_Score']), 3))
```

## 9. Observations

1. The histogram shows the spread of exam scores.
2. Study hours are positively associated with exam scores in this sample.
3. Attendance is positively associated with exam scores in this sample.
4. Average scores differ among study-method categories.
5. The cleaning step removes one duplicate row and imputes one missing value with the median.
6. The data are synthetic, so findings should not be generalized to real students.
